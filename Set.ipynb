```python
# Set Oerations
# UNION
python_students={"shreya","Shubham"}
ml_student={"akash","abhinav","shantnu"}
all_students=python_students.union(ml_student)
all_students=python_students|(ml_student)  # | union

print(all_students)

#
```

```python
# Intersection
python_students={"shreya","Shubham","Vishnu"}
ml_student={"akash","abhinav","shantnu","Vishnu"}
print(python_students.intersection(ml_student))
print(python_students & (ml_student)) # & inersection
```

```python
# Difference
python_students={"shreya","Shubham","Vishnu"}
ml_student={"Vishnu"}
print(python_students.difference(ml_student))
print(python_students-(ml_student)) # - Difference
```

```python
day1={"A","B","C","D"}
day2={"C","D","E","F"}
day3={"E","F","G","H"}


# Find:
# 1.Total Unique Visitors
# 2 Repeat Visitors
# 3.New Visitors on Day3


day1={"A","B","C","D"}
day2={"C","D","E","F"}
day3={"E","F","G","H"}
# 1 Total Uniques Visitors
total_unique=day1|day2|day3
print("Total Unique Visitors:",total_unique)
print("Count:",len(total_unique))

#2 Repeat Visitors
repeat=day1 &day2|day2&day3|day1&day3
print("Repeat Visitors:",repeat)
print("Count:",len(repeat))

# 3 New Visitors on Day3
new_day3=day3-(day1|day2)
print("New Visitors on Day3:",new_day3)
print("Count:",len(new_day3))
```
