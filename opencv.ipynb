```python
image=cv2.imread(r"C:\Users\Vishnu\Desktop\Python_Evening\phot\sachin.jfif")
```

```python
blur=cv2.blur(image,(5,10))
cv2.imshow("Change Pixels",blur)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Gaussian Blur
gaussian=cv2.GaussianBlur(image,(51,51),4)
cv2.imshow("Change Pixels",gaussian)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
median=cv2.medianBlur(image,5)
cv2.imshow("Change Pixels",median)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Bilateral filter
# prereserve Edges
image=cv2.imread(r"C:\Users\Vishnu\Desktop\Python_Evening\phot\sachin.jfif")
bilateral=cv2.bilateralFilter(image,9,75,75)
cv2.imshow("Change Pixels",bilateral)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
0-127
black
128-255
white
```

```python
# Binary Thresold
gray=cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
_,binary=cv2.threshold(gray,127,255,cv2.THRESH_BINARY)
cv2.imshow("Change Pixels",gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Binary Thresold
gray=cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
_,binary=cv2.threshold(gray,127,255,cv2.THRESH_BINARY_INV)
cv2.imshow("Change Pixels",gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
