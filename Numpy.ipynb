```python
import numpy as np
import random
import matplotlib.pyplot as plt

# =========================================================
# TASK SCHEDULING USING REINFORCEMENT LEARNING (Q-LEARNING)
# =========================================================

# -----------------------------
# ENVIRONMENT SETTINGS
# -----------------------------

NUM_SERVERS = 3
NUM_TASKS = 20
EPISODES = 500

ALPHA = 0.1      # Learning Rate
GAMMA = 0.9      # Discount Factor
EPSILON = 0.2    # Exploration Rate

# -----------------------------
# CREATE SERVERS
# -----------------------------

servers = [0 for _ in range(NUM_SERVERS)]

# Example:
# [0,0,0]
# Means all servers have zero load

# -----------------------------
# CREATE Q TABLE
# -----------------------------

# State Space:
# Server loads between 0-20

Q = np.zeros((21, 21, 21, NUM_SERVERS))

# Shape:
# [server1_load][server2_load][server3_load][action]

# -----------------------------
# TRACKING VARIABLES
# -----------------------------

episode_rewards = []
waiting_times = []

# =========================================================
# FUNCTION: GENERATE TASKS
# =========================================================

def generate_task():

    execution_time = random.randint(1, 10)

    priority = random.randint(1, 5)

    task = {
        "execution_time": execution_time,
        "priority": priority
    }

    return task

# =========================================================
# FUNCTION: GET CURRENT STATE
# =========================================================

def get_state(server_loads):

    state = tuple(
        min(load, 20) for load in server_loads
    )

    return state

# =========================================================
# FUNCTION: CHOOSE ACTION
# =========================================================

def choose_action(state):

    # Exploration
    if random.uniform(0,1) < EPSILON:

        return random.randint(0, NUM_SERVERS-1)

    # Exploitation
    else:

        return np.argmax(Q[state])

# =========================================================
# FUNCTION: CALCULATE REWARD
# =========================================================

def calculate_reward(load, execution_time):

    waiting_time = load

    reward = -(waiting_time + execution_time)

    return reward, waiting_time

# =========================================================
# TRAINING LOOP
# =========================================================

for episode in range(EPISODES):

    # Reset Server Loads
    server_loads = [0,0,0]

    total_reward = 0
    total_waiting = 0

    # -----------------------------
    # TASK LOOP
    # -----------------------------

    for _ in range(NUM_TASKS):

        # Generate Task
        task = generate_task()

        execution_time = task["execution_time"]

        # Current State
        state = get_state(server_loads)

        # Choose Server
        action = choose_action(state)

        # Current Load
        current_load = server_loads[action]

        # Calculate Reward
        reward, waiting_time = calculate_reward(
            current_load,
            execution_time
        )

        # Assign Task to Server
        server_loads[action] += execution_time

        # Next State
        next_state = get_state(server_loads)

        # Q-Learning Update
        old_value = Q[state][action]

        next_max = np.max(Q[next_state])

        new_value = old_value + ALPHA * (
            reward + GAMMA * next_max - old_value
        )

        Q[state][action] = new_value

        # Update Metrics
        total_reward += reward
        total_waiting += waiting_time

    # Store Episode Results
    episode_rewards.append(total_reward)

    avg_waiting = total_waiting / NUM_TASKS

    waiting_times.append(avg_waiting)

    # Print Progress
    if episode % 50 == 0:

        print(f"Episode: {episode}")
        print(f"Total Reward: {total_reward}")
        print(f"Average Waiting Time: {avg_waiting:.2f}")
        print("-"*40)

# =========================================================
# TEST TRAINED AGENT
# =========================================================

print("\n==============================")
print("TESTING TRAINED AGENT")
print("==============================")

server_loads = [0,0,0]

for i in range(10):

    task = generate_task()

    state = get_state(server_loads)

    action = np.argmax(Q[state])

    execution_time = task["execution_time"]

    print(f"\nTask {i+1}")
    print(f"Execution Time: {execution_time}")
    print(f"Assigned Server: {action+1}")

    server_loads[action] += execution_time

    print(f"Updated Server Loads: {server_loads}")

# =========================================================
# PLOT 1: REWARD VS EPISODES
# =========================================================

plt.figure(figsize=(10,5))

plt.plot(episode_rewards)

plt.title("Reward vs Episodes")

plt.xlabel("Episodes")

plt.ylabel("Total Reward")

plt.grid(True)

plt.show()

# =========================================================
# PLOT 2: WAITING TIME VS EPISODES
# =========================================================

plt.figure(figsize=(10,5))

plt.plot(waiting_times)

plt.title("Average Waiting Time vs Episodes")

plt.xlabel("Episodes")

plt.ylabel("Waiting Time")

plt.grid(True)

plt.show()
```
