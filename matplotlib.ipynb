```python
fig,axes=plt.subplots(1,3,figsize=(15,5))
for i,j in enumerate(sorted(df['Pclass'].unique())):
    gender_count=df[df['Pclass']==j]['Sex'].value_counts()
    axes[i].pie(
        gender_count,
        autopct='%1.1f%%',
        labels=gender_count.index,
        startangle=90
    )
    axes[i].set_title(f'Pclass {j}')
    print(gender_count)
    print(gender_count.index)
print(fig)
print(axes)
plt.show()
```

```python
pd.crosstab(df['Pclass'],df['Sex']).plot(kind='bar',figsize=(10,5))
plt.legend(title='Sex')
plt.xlabel('Pclass')
plt.ylabel('Count')
plt.title('Count of Passengers by Pclass and Sex')
plt.show()
```

```python

gender_per=pd.crosstab(df['Pclass'],df['Sex'],normalize='index')*100
gender_per.plot(kind='bar',figsize=(10,5),hatch='/',edgecolor='red',color=['lightblue','lightpink'],width=0.8,legend=True)

print(gender_per)
plt.legend(title='Sex')
plt.xlabel('Pclass',color='blue',fontsize=12,fontweight='bold')
plt.ylabel('Count')
plt.title('Count of Passengers by Pclass and Sex')
plt.grid()
plt.show()
```
