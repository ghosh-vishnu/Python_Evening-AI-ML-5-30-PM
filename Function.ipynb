```python
# from os import TMP_MAX
def employee_salary(name,base_salary):
  # Nested Function --Bonus Calculate
  def calculate_bonus():
    return base_salary*0.20
  # Nested Function--Tax Calculate
  def calculate_tax():
    return base_salary*0.12
  bonus=calculate_bonus()
  tax=calculate_tax()
  final_salary=base_salary+bonus-tax
  print(f"Employee Name: {name}")
  print(f"Employee Salary: {base_salary}")
  print(f"Employee Bonus: {bonus}")
  print(f"Tax Deduction: {tax}")
  print(f'Final Salary: {final_salary}')
employee_salary("Vishnu",50000)



```

```python
def decorator_function(display):
  def wrapper():
    print("Before Function")

    display()

    print("After Function")
  return wrapper
def display():
  print("Display Function")
decorated_display=decorator_function(display)
decorated_display()
```

```python
def decorator_function1(display):

  def wrapper1():

    print("Login Success")

    # display()

    print("Logout")

  return wrapper1
# decorator_function1("10")
@decorator_function1
# Dashboad=decorator_function1(Dashboad)
def Dashboad():
  print("Your Welcome! Dashboard is Opend")
display()
```

```python
def zero_division_check(a):
  def validation(num1,num2):
    if num2==0:
      print("cannot divide by zero")
      return
    return a(num1,num2)
  return validation
@zero_division_check
def division(a,b):
  print(a/b)

num1_input=input("enter Your first number")
num2_input=input("enter Your second number")
division(int(num1_input),int(num2_input))
```
