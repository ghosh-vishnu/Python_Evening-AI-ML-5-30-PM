```python
def step(a):
  if len(a)!=10:
    return False
  return (a.count('n'))==a.count('s') and a.count('e')==a.count('w')

print(step(['n','s','n','s','n','s','n','s','n','s']))
```

```python
def step1(n):
  return len(n)==10 and n.count('n')==n.count('s') and n.count('e')==n.count('w')
print(step1(['n','s','n','s','n','s','n','s','n','s']))

```
