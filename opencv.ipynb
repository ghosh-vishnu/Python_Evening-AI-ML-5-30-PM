```python
resized=cv2.resize(image,(300,300))
cv2.imshow("Resized Image",resized)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
small=cv2.resize(image,None,fx=0.5,fy=0.5)
cv2.imshow("Resized Image",small)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
rotated=cv2.rotate(image,cv2.ROTATE_90_CLOCKWISE)
cv2.imshow("90 Image",image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
flip=cv2.flip(image,1)
cv2.imshow("Flip Image",flip)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
gray=cv2.cvtColor(flip,cv2.COLOR_BGR2GRAY)
cv2.imshow("Flip Image",gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
