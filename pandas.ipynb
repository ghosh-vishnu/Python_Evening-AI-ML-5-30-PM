```python
import pandas as pd

data=pd.DataFrame(

    {
        "Name":['Alice','Bob','Charlie','David','Eve'],
        "Age":[25,30,35,40,45],
        "City":['New York','Los Angeles','Chicago','Houston','Phoenix'],
        "Salary":[50000,60000,70000,80000,90000]

    }
)
```

```python
data
```

```python
data[data['Age']>25]
```

```python
data[(data['Age']>25) | (data['Salary']>60000)]
```

```python
data[data['Age'].isin([30,40,50])]
```

```python
data[data['Age'].between(30,40)]
```

```python
data.query('Age > 35')
```
