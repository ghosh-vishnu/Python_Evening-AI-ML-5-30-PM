```python
import pandas as pd
```

```python
df=pd.read_csv("employee_performance_pca_1500.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
df.head()
```

```python
X=df.drop(['Employee_ID','Department'],axis=1)
```

```python
X
```

```python
# Feature Scalling
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
print(X_scaled.shape)
```

```python
X_scaled
```

```python
from sklearn.decomposition import PCA
pca=PCA(n_components=7)
```

```python
pca
```

```python
X_pca=pca.fit_transform(X_scaled)
```

```python
X_pca
```

```python
print(X_pca.shape)
```

```python
pca_df=pd.DataFrame(X_pca,columns=['PC1','PC2','PC3','PC4','PC5','PC6','PC7'])
```

```python
print(pca_df.head())
```

```python
print(sum(pca.explained_variance_ratio_))
```

```python
pca.components_
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,6))
plt.scatter(pca_df['PC1'],pca_df['PC2'])

plt.xlabel("Principal Component of PC1")
plt.ylabel("Principal Component of PC2")

plt.title("PCA Visualization")

plt.show()
```

```python
pca_df['Department']=df['Department']
```

```python
pca_df
```

```python
departments = pca_df["Department"].unique()
plt.figure(figsize=(10,7))
for d in departments:
    temp = pca_df[pca_df["Department"]==d]
    plt.scatter(
        temp["PC1"],
        temp["PC2"],
        label=d
    )
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("Employee PCA by Department")
plt.legend()
plt.show()
```

```python
pca_all=PCA()
```

```python
pca_all = PCA()
X_all = pca_all.fit_transform(X_scaled)
```

```python
plt.figure(figsize=(8,5))

plt.plot(
range(1,len(pca_all.explained_variance_ratio_)+1),
pca_all.explained_variance_ratio_,
marker='o'
)
plt.xlabel("Principal Components")
plt.ylabel("Explained Variance Ratio")
plt.title("Scree Plot")
plt.grid(True)

plt.show()
```

```python
pca_df.to_csv(
"employee_pca_output.csv",
index=False
)
```
